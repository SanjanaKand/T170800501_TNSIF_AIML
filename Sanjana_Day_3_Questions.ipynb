```python
# 1 :Find the largest of three numbers.
def Summation(Num1 ,Num2 , Num3):
  return max(Num1,Num2,Num3)

def main():
  No1 = int(input("Enter first number :"))
  No2 = int(input("Enter second number :"))
  No3 = int(input("Enter third number :"))

  Ret = Summation(No1 , No2 , No3)

  print(f"The maximum number from these numbers {No1} , {No2} , {No3} is : {Ret}")

if __name__ == "__main__":
  main()
```

```python
# 4 : Find the factorial of a number using a loop.
def Factorial(No):
  fact = 1

  for i in range(1,No+1):
    fact = fact * i
  return fact

def main():
  Number = int(input("Enter a number :"))

  Ret = Factorial(Number)

  print(f"Factorial of {Number} is : {Ret}")

if __name__ == "__main__":
  main()
```

```python
# 7 : Reverse a number using a loop.
def Reverse(No):
  string_var = str(No)
  ans = string_var[::-1]
  return ans

def main():
  Number = int(input("Enter a number containing more than 2 digits :"))

  Ret = Reverse(Number)

  print(f"The reverse of {Number} is {Ret}")

if __name__ == "__main__":
  main()
```

```python
# 8 : Write a function to check if a number is palindrome.

def Palindrome(No):
    return str(No) == str(No)[::-1]

def main():
    Number = int(input("Enter a number: "))

    if Palindrome(Number):
        print(f"{Number} is a palindrome.")
    else:
        print(f"{Number} is not a palindrome.")

if __name__ == "__main__":
    main()
```

```python
# 9 : Check whether a number is prime.

def Prime(No):
    if No <= 1:
        return False

    for i in range(2, No):
        if No % i == 0:
            return False

    return True

def main():
    Number = int(input("Enter a number: "))

    if Prime(Number):
        print(f"{Number} is a prime number.")
    else:
        print(f"{Number} is not a prime number.")

if __name__ == "__main__":
    main()
```

```python
# 10 : Write a function to calculate sum of digits of a number.
def SumDigits(No):
    Sum = 0

    for Digit in str(No):
        Sum = Sum + int(Digit)

    return Sum

def main():
    Number = int(input("Enter a number: "))

    Ret = SumDigits(Number)

    print(f"Sum of digits is: {Ret}")

if __name__ == "__main__":
    main()
```

```python
# 12 : Find the largest and smallest element in a list.
def main():
    Numbers = [10, 45, 2, 89, 34, 17]

    Largest = Numbers[0]
    Smallest = Numbers[0]

    for i in Numbers:
        if i > Largest:
            Largest = i

        if i < Smallest:
            Smallest = i

    print("List:", Numbers)
    print("Largest element:", Largest)
    print("Smallest element:", Smallest)

if __name__ == "__main__":
    main()
```

```python
# 13 : Remove duplicates from a list.
def main():
    Numbers = [10, 20, 30, 20, 40, 10, 50]

    NewList = []

    for i in Numbers:
        if i not in NewList:
            NewList.append(i)

    print("Original List:", Numbers)
    print("List after removing duplicates:", NewList)

if __name__ == "__main__":
    main()
```

```python
# 14 : Reverse a list without using built-in functions.
def main():
    Numbers = [10, 20, 30, 40, 50]

    ReverseList = []

    for i in range(len(Numbers)-1, -1, -1):
        ReverseList.append(Numbers[i])

    print("Original List:", Numbers)
    print("Reversed List:", ReverseList)

if __name__ == "__main__":
    main()
```

```python
# 23. Count Frequency of Characters in a String
def main():
    String = input("Enter a string: ")

    Frequency = {}

    for Ch in String:
        if Ch in Frequency:
            Frequency[Ch] += 1
        else:
            Frequency[Ch] = 1

    print(Frequency)

if __name__ == "__main__":
    main()
```

```python
# 27. Create a Class Rectangle
class Rectangle:
    def __init__(self, Length, Width):
        self.Length = Length
        self.Width = Width

    def Area(self):
        return self.Length * self.Width

    def Perimeter(self):
        return 2 * (self.Length + self.Width)

def main():
    Obj = Rectangle(10, 5)

    print("Area:", Obj.Area())
    print("Perimeter:", Obj.Perimeter())

if __name__ == "__main__":
    main()
```

```python
# 28. Create a Class Employee
class Employee:
    def __init__(self, Salary):
        self.Salary = Salary

    def IncreaseSalary(self, Amount):
        self.Salary = self.Salary + Amount

    def Display(self):
        print("Salary:", self.Salary)

def main():
    Obj = Employee(50000)

    Obj.Display()

    Obj.IncreaseSalary(5000)

    Obj.Display()

if __name__ == "__main__":
    main()
```

```python
# 29. Create a Class ShoppingCart
class ShoppingCart:
    def __init__(self):
        self.Cart = {}

    def AddItem(self, Item, Price):
        self.Cart[Item] = Price

    def TotalBill(self):
        return sum(self.Cart.values())

def main():
    Obj = ShoppingCart()

    Obj.AddItem("Pen", 20)
    Obj.AddItem("Book", 150)
    Obj.AddItem("Bag", 500)

    print("Cart:", Obj.Cart)
    print("Total Bill:", Obj.TotalBill())

if __name__ == "__main__":
    main()
```

```python
# 30. Create a Class Library
class Library:
    def __init__(self):
        self.Books = []

    def AddBook(self, Book):
        self.Books.append(Book)

    def RemoveBook(self, Book):
        if Book in self.Books:
            self.Books.remove(Book)

    def Display(self):
        print("Books:", self.Books)

def main():
    Obj = Library()

    Obj.AddBook("Python")
    Obj.AddBook("Java")
    Obj.AddBook("C++")

    Obj.Display()

    Obj.RemoveBook("Java")

    Obj.Display()

if __name__ == "__main__":
    main()
```
